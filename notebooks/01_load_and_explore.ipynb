```python
import duckdb

con = duckdb.connect("../data/fraud.duckdb")
con.execute("""
    CREATE OR REPLACE TABLE train AS
    SELECT *
    FROM read_csv('../data/train_transaction.csv') t
    LEFT JOIN read_csv('../data/train_identity.csv') i
      USING (TransactionID)
""")
con.execute("""
    SELECT COUNT(*) AS rows, ROUND(AVG(isFraud) * 100, 2) AS fraud_pct
    FROM train
""").df()
```

```python
con.execute("""
    SELECT ProductCD,
           COUNT(*) AS transactions,
           ROUND(AVG(isFraud) * 100, 2) AS fraud_pct
    FROM train
    GROUP BY ProductCD
    ORDER BY fraud_pct DESC
""").df()
```
